# 1. Obtener el Dataset de Roboflow

```python
from roboflow import Roboflow

rf = Roboflow(api_key="") # USE YOUR API KEY
project = rf.workspace("roboflow-jvuqo").project("football-players-detection-3zvbc")
version = project.version(1)
dataset = version.download("yolov5")

print(f"\nDataset descargado correctamente en la ruta: {dataset.location}")
```

# 2. Entrenamiento del Modelo

```python
from ultralytics import YOLO

# 1. Cargar el modelo YOLOv5x (Extra Large) preentrenado
model = YOLO("yolov5xu.pt") 

# 2. Iniciar el entrenamiento usando la API de Python
results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=100,
    imgsz=640,
    project="entrenamiento_futbol", # Carpeta base para guardar resultados
    name="modelo_yolov5x"           # Subcarpeta para este entrenamiento específico
)
```
